```python
df['PassengerId'] = df['PassengerId'].astype(int)
df['Survived'] = df['Survived'].astype('category')
```

```python
df['PassengerId_norm'] = (df['PassengerId'] - df['PassengerId'].min()) / (df['PassengerId'].max() - df['PassengerId'].min())

print(df.head())
```

```python
df['Survived_numeric'] = df['Survived'].cat.codes
print(df.head())
```
